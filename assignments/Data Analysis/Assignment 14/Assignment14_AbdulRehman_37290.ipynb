# - - - Dataset 1: Pakistani E-Commerce Data - - - 

```python
import pandas as pd
```

```python
df = pd.read_csv("datasets/pakistani_ecommerce_dataset.csv")
```

```python
df.info()
```

```python
df.head()
```

Q1: Which is the most popular product category? 


```python
df["product_category"].value_counts()
```

Ans) Home Decor is the most popular category.

Q2: What is the average order value? 


```python
average = df["order_value"].mean()
print("Average Order Amont:", round(average, 2))
```

Ans) Average Order Valuee is Rs 2832.15.

Q3: Which city has the highest number of orders? 

```python
df["city"].value_counts()
```

Ans) Karachi has the highest number of orders.

Q4: What is the most used payment method?

```python
df["payment_method"].value_counts()
```

Ans) "Cash on delivery" payment method was used the most.

Q5: Draw a bar chart showing average order value by city. 


```python
import matplotlib.pyplot as plt

city_avg = df.groupby("city")["order_value"].mean()

print(round(city_avg,2))
```

```python
city_avg.plot(kind="bar", figsize=(9, 5))

plt.title("Average Order Value by City")
plt.xlabel("City")
plt.ylabel("Average Order Value (PKR)")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

# - - - Dataset 2: Student Lifestyle & Study Habits - - -

```python
df2 = (pd.read_csv("datasets/student_lifestyle_dataset.csv"))
```

```python
df2.info()
```

```python
df2.head()
```

Q1: What is the average number of study hours? 

```python
avg = df2["study_hours"].mean()
print("Average Study Hours =",round(avg,2))
```

Ans) Average Study hours is 4.4 hours.

Q2: Who sleeps more on average – males or females?

```python
sleep_avg = df2.groupby("gender")["sleep_hours"].mean()

print(round(sleep_avg,2))
```

Ans) According to this dataset, on average, females sleep more.
    

Q3: Is there a relationship between screen time and sleep hours? 


```python
correlation = df2["screen_time"].corr(df2["sleep_hours"])

print(f"Correlation: {correlation:.2f}")
```

Ans) There is almost no linear relationship between screen time and sleep hours.

Q4: Draw a histogram of daily transport cost.

```python
plt.figure(figsize=(8, 5))

plt.hist(df2["daily_transport_cost"], bins=10)

plt.title("Distribution of Daily Transport Cost")
plt.xlabel("Daily Transport Cost")
plt.ylabel("Frequency")

plt.show()
```

Q5: Compare average internet usage by gender. 

```python
avg2 = df2.groupby("gender")["internet_usage"].mean()
print("Avg Internet Usage =",round(avg2,2))
```

Ans) On average, internet usage is more for males than females. The difference is 0.28.
