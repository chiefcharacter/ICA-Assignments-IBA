Q1: What is the most preferred tea type?

Q2: What is the average number of cups per gender?

```python
avg_cups = df.groupby("gender")["cups_per_day"].mean()
print("Average Cups/Gender =", round(avg_cups, 2))
```

Ans) Average Number of Cups for Males = 3 cups/day and Females = 2.6 cups/day.

Q3: Make a bar chart of snack choices.

```python
import matplotlib.pyplot as plt
```

```python
snack = df["snack_choice"].value_counts()
```

```python
snack.plot(kind="bar")

plt.title("Snack Choices")
plt.xlabel("Snacks")
plt.ylabel("Number of Students")
plt.show()
```

Q4: Compare tea preference by age group

```python
tea_pref = pd.crosstab(df["age_group"], df["prefers_tea"])
print(tea_pref)
```

Ans) Among students of age group 18-20, 20 students prefer tea while 19 students do not. 

In the 21–23 age group, 17 students do not prefer tea compared to 15 who do. Among those aged 24–26, 16 prefer tea while 13 do not. Overall, tea preference is relatively balanced across all three age groups, with slight differences between them.
