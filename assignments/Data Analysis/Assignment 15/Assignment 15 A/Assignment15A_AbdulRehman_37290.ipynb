# Ride-Hailing in Pakistan

```python
import pandas as pd
```

```python
df = pd.read_csv("dataset/ride_hailing_dataset.csv")
```

```python
df.head()
```

Q1: What is the average fare by service type?


```python
avg = df.groupby("service_type")["fare"].mean()
print(round(avg,2))
```

Ans) Average Fare Price by Service Type:
Careem = Rs. 544.98
InDrive = Rs. 573.51
Rickshaw = Rs. 486.64
Uber = Rs. 486.13

Q2: Which city has the most trips?

```python
df["city"].value_counts()
```

Ans) Islamabad has the most trips.

Q3: What is the average distance per trip?


```python
distance = df["distance_km"].mean()
print("Average Distance/Trip =", round(distance,2))
```

Ans) Average distance per trip is 9.61 kilometers.

Q4: Compare trip ratings by gender.


```python
rating = df.groupby("gender")["trip_rating"].mean()
print("Trip Rating =",round(rating,2))
```

Ans) Trip rating for males was higher compared to females.
