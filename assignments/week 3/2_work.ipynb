```python
oil = clean[(clean["item"] == "Oil B") & (clean["city"] == "City 1")]

oil = oil.sort_values("date")

first = oil["price"].dropna().iloc[0]
last = oil["price"].dropna().iloc[-1]

print("Percentage increase:", round(((last-first)/first)*100, 2), "%")

oil = clean[(clean["item"] == "Oil B") & (clean["city"] == "City 2")]

oil = oil.sort_values("date")

first = oil["price"].dropna().iloc[0]
last = oil["price"].dropna().iloc[-1]

print("Percentage increase:", round(((last-first)/first)*100, 2), "%")
```
